```python
#Email Validation using Regex
import re
email=input("Enter Email:")
pattern=r'^[A-Za-z0-9_%+-]+@[A-Za-z0-9-]+\.[A-Za-z]{2,}$'
if re.fullmatch(pattern, email):
 print("Valid Email")
else:
 print("Invalid Email")
```
